## Why Text Vectorization is Required in NLP

In Natural Language Processing (NLP), we deal with human language, which is inherently textual. Computers, however, cannot directly process text—they require **numerical input**. Text vectorization bridges this gap by converting words or sentences into numerical vectors. This is essential because:

- Machine learning algorithms operate on numbers, not raw text.  
- It enables tasks like classification, clustering, sentiment analysis, and translation.  
- Vectorization allows models to capture patterns, relationships, and statistical properties of language.  

---


## Key Concepts in NLP

#### 1. Corpus
A **corpus** is a large and structured collection of texts used for linguistic analysis or training NLP models.  
- Example: A collection of movie reviews, news articles, or social media posts.  
- It serves as the dataset from which vocabulary and patterns are extracted.  

#### 2. Vocabulary
The **vocabulary** is the set of all unique words present in the corpus.  
- Example: If the corpus contains sentences like *"people watch dhurandar"* and *"people write comment"*, the vocabulary would be:  
  `["people", "watch", "dhurandar", "write", "comment"]`  
- Vocabulary size directly impacts the dimensionality of vector representations.  

#### 3. Document
A **document** refers to a single text entry within the corpus.  
- Example: `"people watch dhurandar"` is one document.  
- In NLP tasks, each document is often treated as a unit for classification, clustering, or retrieval.  

#### 4. Word
A **word** is the smallest meaningful unit of text in the vocabulary.  
- Example: `"people"`, `"watch"`, `"dhurandar"`.  
- Words are the building blocks of documents and are mapped into numerical vectors during text vectorization.  


## One-Hot Encoding (OHE) Example
**Dataset:**

| text                        | label |
|-----------------------------|-------|
| people watch dhurandar      | 1     |
| dhuandar watch dhurandar    | 1     |
| people write comment        | 0     |
| dhurandar write comment     | 0     |

**Step 1: Build Vocabulary**  
Unique words:  
`["people", "watch", "dhurandar", "dhuandar", "write", "comment"]`

**Step 2: Encode Sentences**  
Each sentence is represented as a binary vector (1 if word present, 0 if absent):

| text                        | OHE Vector                  | label |
|-----------------------------|-----------------------------|-------|
| people watch dhurandar      | [1, 1, 1, 0, 0, 0]          | 1     |
| dhuandar watch dhurandar    | [0, 1, 1, 1, 0, 0]          | 1     |
| people write comment        | [1, 0, 0, 0, 1, 1]          | 0     |
| dhurandar write comment     | [0, 0, 1, 0, 1, 1]          | 0     |

---

## Pros and Cons of One-Hot Encoding

### Pros
- Simple and easy to implement.  
- Preserves exact word identity (no ambiguity).  
- Works well for small vocabularies and basic classification tasks.  

### Cons
- Produces sparse vectors (many zeros → memory inefficiency).  
- Vocabulary size grows quickly → large dimensionality.  
- No semantic meaning captured (e.g., "watch" ≠ "see").  
- Cannot handle unseen words (out-of-vocabulary problem).  


# Bag of Words (BoW) with CountVectorizer

## What is Bag of Words?
- A simple way to represent text as numbers.  
- Each unique word in the dataset becomes a column (feature).  
- Each row (sentence) is represented by **counts** of those words.  

Example:  

Sentence: *"people watch dhurandar"*  
BoW: {"people":1, "watch":1, "dhurandar":1, "write":0, "comment":0}  

---

##  Dataset
We will use the same dataset as before:

| text                        | label |
|-----------------------------|-------|
| people watch dhurandar      | 1     |
| dhurandar watch dhurandar   | 1     |
| people write comment        | 0     |
| dhurandar write comment     | 0     |

---

## Vocabulary

Unique words extracted:  
`["people", "watch", "dhurandar", "write", "comment"]`

---

## BoW Representation
Each sentence is represented by the **frequency count** of words:

| text                        | BoW Vector             | label |
|-----------------------------|------------------------|-------|
| people watch dhurandar      | [1, 1, 1, 0, 0]        | 1     |
| dhurandar watch dhurandar   | [0, 1, 2, 0, 0]        | 1     |
| people write comment        | [1, 0, 0, 1, 1]        | 0     |
| dhurandar write comment     | [0, 0, 1, 1, 1]        | 0     |

---

## Difference Between OHE and BoW
- **OHE (One-Hot Encoding):**  
  - Only indicates presence (1) or absence (0) of a word.  
  - Example: `"dhurandar watch dhurandar"` → [0, 1, 1, 1, 0, 0]  

- **BoW (Bag of Words):**  
  - Captures the **frequency** of words in a sentence.  
  - Example: `"dhurandar watch dhurandar"` → [0, 1, 2, 0, 0]  

---

## Pros and  Cons of Bag of Words

### Pros
- Simple and easy to implement.  
- Captures word frequency, which can be useful for text classification.  
- Works well for small datasets and basic NLP tasks.  

### Cons
- Still produces sparse vectors for large vocabularies.  
- Ignores word order and context (semantic meaning is lost).  
- Vocabulary size grows quickly with larger corpora.  
- Cannot handle unseen words (out-of-vocabulary problem).

```python
# --------------------------------------------
# 1. Import libraries
# --------------------------------------------
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# --------------------------------------------
# 2. Create dataset
# --------------------------------------------
data = {
    "text": [
        "people watch dhurandar",
        "dhurandar watch dhurandar",
        "people write comment",
        "dhurandar write comment"
    ],
    "label": [1, 1, 0, 0]
}

df = pd.DataFrame(data)
print(df)
```

### Step 3: Apply Bag of Words

We use `CountVectorizer` to convert text into numerical vectors.

## 3. Bag of Words using CountVectorizer

```python
pip install scikit-learn
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer()
bow = cv.fit_transform(df["text"])
```

```python
# Vocabulary (unique words mapped to column index)
print("Vocabulary:", cv.vocabulary_)
```

### Step 4: See Vectorized Format
Each row corresponds to a sentence. Each column corresponds to a word.

```python
# Convert sparse matrix to array for visualization
print("BoW Array:\n", bow.toarray())

# Show feature names (columns)
print("Feature Names:", cv.get_feature_names_out())
```

### Step 5: Test Prediction with New Sentence

Now let’s transform a new sentence:

```python
# --------------------------------------------
# 5. Transform new sentence
# --------------------------------------------
new_sentence = ["dhurandar watch and write a comment of dhurandar"]
vectorized = cv.transform(new_sentence).toarray()
print("Vectorized New Sentence:\n", vectorized)
```

## Explanation of Results

- Words like **"and"** and **"of"** were **not in the training data**.  
- CountVectorizer simply **ignores them** — they don’t appear in the vocabulary, so they don’t affect the vector.  
- This is why there’s **no out‑of‑vocabulary error**: unseen words are dropped silently.

---

### Parameter: `max_features`
- Limits the number of words in the vocabulary.  
- Example: `CountVectorizer(max_features=5)` → only the **top 5 most frequent words** are kept.  
- Useful when you have huge text data and want to reduce dimensionality.  

If `max_features=5`, rare words are ignored, and only the most common ones are represented in the BoW.

# One‑Hot Encoding (OHE) : Limitations
- **Sparsity** → Each word is represented as a long vector with mostly zeros. For large vocabularies, this wastes memory.  
- **No fixed size** → Vocabulary grows with new words, so vector length keeps changing.  
- **Out‑of‑Vocabulary (OOV)** → If a new word appears that wasn’t in training, OHE cannot represent it.  
- **No semantic meaning** → Each word is independent; “dog” and “puppy” are unrelated in OHE.  

---

## Bag of Words (BoW) – Improvements
BoW solves many of these issues:

### Fixed Vocabulary Size
- When you fit `CountVectorizer`, it builds a fixed dictionary of words from the training data.  
- Every sentence is represented using this same dictionary → consistent vector size.  

### Handles OOV Gracefully
- New words (like “and”, “of”) are simply ignored.  
- They don’t break the model or cause errors, they just don’t contribute to the vector.  

### Captures Frequency Information
- Unlike OHE (just 0/1), BoW counts how many times a word appears.  
- Example: *“dhurandar dhurandar watch”* → {"dhurandar":2, "watch":1}.  
- This gives richer information about word importance.  

### Less Sparse than OHE
- Still sparse for large vocabularies, but more compact than OHE because it uses counts instead of huge binary vectors.  

### Foundation for Advanced Models
- BoW is the stepping stone to **TF‑IDF** (weights words by importance) and **embeddings** (capture meaning).  
- Students see how text → numbers → ML pipeline works.  

---

### Example Recap 
Sentence:  
*"dhurandar watch and write a comment of dhurandar"*  

Vocabulary (from training):  
`["people", "watch", "dhurandar", "write", "comment", "comman", "dhuandar"]`  

Vector: `[0, 1, 2, 1, 1, 0, 0]`  

- “and” and “of” are ignored (not in vocabulary).  
- “dhurandar” counted twice.  
- No OOV error.  

---

## Parameter: `max_features`
- Limits vocabulary size to the **top N most frequent words**.  
- Example: `CountVectorizer(max_features=5)` → keeps only the 5 most common words.  

### Useful When:
- Dataset is huge.  
- You want to reduce dimensionality.  
- You want to avoid rare words that add noise.  

Teaching Tip: Show students how `max_features=3` changes the vocabulary and vectors, so they see the trade‑off between **information vs simplicity**.  

---

## Key Insight for Students
- **OHE** is too rigid and inefficient.  
- **BoW** is more practical: fixed size, ignores unknown words safely, and captures word counts.  
- It’s the bridge between raw text and more advanced representations like **TF‑IDF** and **embeddings**.  

## N‑grams 

### What are N‑grams?
- **N‑gram** = sequence of N words.
- **Unigram** → single words (like Bag of Words).
- **Bigram** → pairs of consecutive words.
- **Trigram** → triplets of consecutive words.

## Why use N‑grams?
- They capture **context** and **word order**.
- Example:  
  - Unigram: "watch", "dhurandar"  
  - Bigram: "watch dhurandar" (captures the phrase)

---

##  Dataset

We use the same dataset:

| text                          | label |
|-------------------------------|-------|
| people watch dhurandar        | 1     |
| dhuandar watch dhurandar      | 1     |
| people write comment          | 0     |
| dhurandar write comman        | 0     |

```python
# --------------------------------------------
# 1. Import libraries
# --------------------------------------------
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# --------------------------------------------
# 2. Create dataset (corrected)
# --------------------------------------------
data = {
    "text": [
        "people watch dhurandar",
        "dhurandar watch dhurandar",
        "people write comment",
        "dhurandar write comment"
    ],
    "label": [1, 1, 0, 0]
}

df = pd.DataFrame(data)
print(df)
```

## Step 3: Apply Bi‑gram Bag of Words

We set `ngram_range=(2,2)` → this means **only bigrams** will be considered.

```python
cv_bigram = CountVectorizer(ngram_range=(2,2))
bow_bigram = cv_bigram.fit_transform(df["text"])
```

```python
# Vocabulary (unique bigrams mapped to column index)
print("Bigram Vocabulary:", cv_bigram.vocabulary_)
```

##  Step 4: See Vectorized Format

Each row corresponds to a sentence. Each column corresponds to a bigram.

```python
# Convert sparse matrix to array for visualization
print("BoW Bigram Array:\n", bow_bigram.toarray())

# Show feature names (columns)
print("Feature Names:", cv_bigram.get_feature_names_out())
```

## Explanation of Results

- Vocabulary now contains **pairs of words** instead of single words.  
- Example bigrams from our dataset:
  - "people watch"
  - "watch dhurandar"
  - "dhuandar watch"
  - "people write"
  - "write comment"
  - "dhurandar write"
  - "write comman"

Notice how **order matters**:  
- "watch dhurandar" ≠ "dhurandar watch".  
- This helps capture context that plain Bag of Words misses.

---

## Key Insight
- **Unigram BoW** → captures word presence/frequency.  
- **Bigram BoW** → captures short phrases and word order.  
- Useful for tasks like sentiment analysis, spam detection, or phrase‑based classification.

### Advantages & Disadvantages of N‑grams

#### Advantages of N‑grams
1. **Captures word order**  
   - Unlike Bag of Words (unigrams), N‑grams preserve short context.  
   - Example:  
     - Unigram: "watch", "dhurandar"  
     - Bigram: "watch dhurandar" → captures the phrase.

2. **Improves meaning in classification**  
   - Sentiment analysis:  
     - "not good" → bigram shows negative sentiment.  
     - Unigram "good" alone would mislead.

3. **Simple to implement**  
   - Extends Bag of Words easily with `ngram_range`.  
   - No complex math required.

---

#### Disadvantages of N‑grams
1. **Vocabulary explosion**  
   - More N‑grams → larger vocabulary → higher dimensionality.  
   - Example:  
     - Unigrams: 5 words → 5 features.  
     - Bigrams: 5 words → up to 10 features.  
     - Trigrams: grows even faster.

2. **Sparsity increases**  
   - Many N‑grams appear rarely, leading to sparse vectors.  
   - Example: "people write comment" → bigram "write comment" appears only once.

3. **Limited context**  
   - N‑grams only capture short phrases.  
   - They don’t understand long‑range dependencies (e.g., meaning across sentences).

4. **Computational cost**  
   - Larger vocabulary → more memory and slower training.  
   - Not scalable for very large corpora.

---

## Key Insight
- **N‑grams add context** compared to plain Bag of Words.  
- But they also add **complexity and sparsity**.  
- They are a stepping stone toward more advanced techniques like **TF‑IDF** and **word embeddings**, which balance context and efficiency.

# TF‑IDF (Term Frequency – Inverse Document Frequency)

## Motivation
- Bag of Words treats all words equally.  
- But some words are **common everywhere** (like "the", "and") → not very informative.  
- TF‑IDF highlights **rare but important words** in a document.  
- Example: If "dhurandar" appears often in one document but rarely in the whole corpus, TF‑IDF gives it **higher weight**.

---

##  Formula
- **TF (t, d)** = (Number of times term *t* appears in document *d*) ÷ (Total terms in document *d*)  
- **IDF (t)** = log( Total number of documents ÷ Number of documents containing term *t* )  
- **TF‑IDF (t, d)** = TF(t, d) × IDF(t)

---

##  Dataset
We use 4 documents:

- d1: "people watch dhurandar"  
- d2: "dhurandar watch dhurandar"  
- d3: "people write comment"  
- d4: "dhurandar write comment"  

Vocabulary: **people, watch, dhurandar, write, comment**

---

##  Step 1: Term Frequency (TF)
Examples:
- TF(people, d1) = 1/3  
- TF(dhurandar, d2) = 2/3  

---

## Step 2: Inverse Document Frequency (IDF)
- Total documents = 4  
- IDF(dhurandar) = log(4/3)  
- IDF(people) = log(4/2) = log(2)  
- IDF(watch) = log(4/2) = log(2)  
- IDF(write) = log(4/2) = log(2)  
- IDF(comment) = log(4/2) = log(2)  



We have 4 documents in total.

- **IDF(dhurandar)** = log(4/3) ≈ **0.125**  
- **IDF(people)** = log(4/2) = log(2) ≈ **0.693**  
- **IDF(watch)** = log(4/2) = log(2) ≈ **0.693**  
- **IDF(write)** = log(4/2) = log(2) ≈ **0.693**  
- **IDF(comment)** = log(4/2) = log(2) ≈ **0.693**

---

###  Interpretation
- **dhurandar** appears in 3 out of 4 documents → IDF is small (≈0.125). It’s common, so less informative.  
- **people, watch, write, comment** each appear in 2 out of 4 documents → IDF is moderate (≈0.693). They are less common than *dhurandar*, so more informative.  
- If a word appeared in **all 4 documents**, its IDF would be log(4/4) = log(1) = **0** → meaning it carries no distinguishing power.  
- If a word appeared in **only 1 document**, its IDF would be log(4/1) = log(4) ≈ **1.386** → meaning it’s rare and highly informative.

---
Interpretation:
- Words appearing in **many documents** → lower IDF → less informative.  
- Words appearing in **few documents** → higher IDF → more informative.

---

## Step 3: TF‑IDF Table (Easy Interpretation)

| Word      | d1 (TF×IDF) | d2 (TF×IDF) | d3 (TF×IDF) | d4 (TF×IDF) |
|-----------|-------------|-------------|-------------|-------------|
| people    | (1/3 × log2) | (0 × log2)  | (1/3 × log2) | (0 × log2)  |
| watch     | (1/3 × log2) | (1/3 × log2) | (0 × log2)  | (0 × log2)  |
| dhurandar | (1/3 × log(4/3)) | (2/3 × log(4/3)) | (0 × log(4/3)) | (1/3 × log(4/3)) |
| write     | (0 × log2)  | (0 × log2)  | (1/3 × log2) | (1/3 × log2) |
| comment   | (0 × log4)  | (0 × log4)  | (1/3 × log4) | (0 × log4)  |




## (Numeric Values)

We use ln (natural log) for IDF.

- log(2) ≈ 0.693  
- log(4/3) ≈ 0.288  
- log(4) ≈ 1.386  

### TF‑IDF Table

| Word      | d1 (TF×IDF) | d2 (TF×IDF) | d3 (TF×IDF) | d4 (TF×IDF) |
|-----------|-------------|-------------|-------------|-------------|
| people    | 1/3 × 0.693 ≈ **0.231** | 0 | 1/3 × 0.693 ≈ **0.231** | 0 |
| watch     | 1/3 × 0.693 ≈ **0.231** | 1/3 × 0.693 ≈ **0.231** | 0 | 0 |
| dhurandar | 1/3 × 0.288 ≈ **0.096** | 2/3 × 0.288 ≈ **0.192** | 0 | 1/3 × 0.288 ≈ **0.096** |
| write     | 0 | 0 | 1/3 × 0.693 ≈ **0.231** | 1/3 × 0.693 ≈ **0.231** |
| comment   | 0 | 0 | 1/3 × 1.386 ≈ **0.462** | 1/3 × 1.386 ≈ **0.462** |

---

## Interpretation
- **people, watch, write** → moderate weights (~0.231) because they appear in 2 documents.  
- **dhurandar** → very low weight (~0.096–0.192) because it appears in 3 documents (quite common).  
- **comment** → highest weight (~0.462) because it’s rarer and appears in fewer documents.  

TF‑IDF highlights **rare but meaningful words** (like *comment*) and downplays **common words** (like *dhurandar*).


| Word      | d1 | d2 | d3 | d4 |
|-----------|----|----|----|----|
| people    | appears once → medium importance | absent | appears once → medium importance | absent |
| watch     | appears once → lower importance (common word) | appears once → lower importance | absent | absent |
| dhurandar | appears once → small importance (quite common) | appears twice → slightly higher weight | absent | appears once → small importance |
| write     | absent | absent | appears once → medium importance | appears once → medium importance |
| comment   | absent | absent | appears once → high importance (rare word) | appears once → high importance |

---




### Key Insight
- **Common words** (like *watch*, *dhurandar*) → lower TF‑IDF weight because they appear in many documents.  
- **Rare words** (like *comment*) → higher TF‑IDF weight because they appear in fewer documents.  
- TF‑IDF = **frequency in a document × rarity across all documents**.  

Think of it like this:  
- **BoW** just counts words.  
- **TF‑IDF** asks: *“Is this word special compared to the whole dataset?”*
