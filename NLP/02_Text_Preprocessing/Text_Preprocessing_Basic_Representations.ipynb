# NLP – Text Preprocessing

## Why Text Preprocessing Matters
- Raw text data is often noisy, inconsistent, and unstructured.  
- Preprocessing prepares text so that machine learning and deep learning models can understand it.  
- It reduces complexity, removes irrelevant information, and standardizes input for better performance.

---

## Common Steps in Text Preprocessing

### 1. **Tokenization**
- Splitting text into smaller units (words, subwords, or characters).  
- Example: `"Deep Learning is powerful"` → `["Deep", "Learning", "is", "powerful"]`.

### 2. **Lowercasing**
- Converting all text to lowercase to ensure consistency.  
- Example: `"TensorFlow"` → `"tensorflow"`.

### 3. **Removing Punctuation and Special Characters**
- Eliminates symbols that don’t contribute to meaning.  
- Example: `"Hello!!!"` → `"Hello"`.

### 4. **Stopword Removal**
- Stopwords are common words (like *the, is, and*) that may not add value.  
- Removing them reduces noise in the dataset.

### 5. **Stemming and Lemmatization**
- **Stemming:** Reduces words to their root form (e.g., `"running"` → `"run"`).  
- **Lemmatization:** Converts words to their dictionary form (e.g., `"better"` → `"good"`).  
- Helps group similar words together.

### 6. **Handling Numbers**
- Numbers can be removed, normalized, or converted into words depending on context.  
- Example: `"100"` → `"one hundred"`.

### 7. **Vectorization**
- Converting text into numerical form for model input.  
- Methods include Bag of Words, TF‑IDF, Word2Vec, and embeddings.

---

## Key Insight
Text preprocessing is the **foundation of NLP pipelines**.  
It transforms raw text into a clean, structured, and standardized format, enabling models to focus on meaningful patterns rather than noise.  
Without preprocessing, models struggle with inconsistencies and irrelevant details in language data.

# Understanding the Corpus

##  What is a Corpus?
- A **corpus** is a large and structured collection of texts (documents, sentences, or words).  
- It serves as the **raw material** for Natural Language Processing (NLP) tasks.  
- Think of it as the "dataset of language" that models learn from.

---

## How is a Corpus Formed?
- Texts are gathered from various sources:
  - Books, newspapers, articles
  - Social media posts, blogs, forums
  - Transcribed speech or dialogue
- The collected texts are cleaned, standardized, and stored in a structured format.  
- Corpora can be **general-purpose** (covering everyday language) or **domain-specific** (medical, legal, technical).

---

## How Does a Corpus Work in NLP?
- Provides the **training data** for models to learn vocabulary, grammar, and semantics.  
- Used for:
  - Building frequency counts of words
  - Training embeddings (Word2Vec, GloVe, BERT)
  - Tokenization and stopword lists
  - Statistical analysis of language patterns
- Corpus size and diversity directly affect model performance.

---

## Why is a Corpus Important in NLP?
- **Foundation for preprocessing:** Tokenization, stopword removal, stemming, and lemmatization rely on corpus data.  
- **Vocabulary building:** Defines which words/subwords are recognized by the model.  
- **Context learning:** Helps models understand meaning, relationships, and usage of words.  
- **Evaluation:** Corpora are also used as benchmarks (e.g., Penn Treebank, Brown Corpus, Wikipedia dumps).

---

##  Key Insight
A corpus is the **backbone of NLP**.  
Without a well-structured corpus, models cannot learn language patterns effectively.  
The quality, size, and relevance of the corpus determine how well an NLP system performs in real-world tasks.

````python
import nltk
from nltk.corpus import stopwords
import re
````

````python
# Download stopwords (only needed once)
nltk.download("stopwords")
````

````python
# English stopwords list
stop_words = set(stopwords.words("english"))

# Example sentences
sentence1 = "This is an example showing how stopword removal works."
sentence2 = "Natural Language Processing is fun and very powerful."
sentence3 = "Stopwords are common words that may not add much meaning."
````

````python
# --------------------------------------------
# 1. Tokenize sentences into words
# --------------------------------------------
def tokenize(text):
    # Lowercase + remove punctuation
    text = re.sub(r"[^\w\s]", "", text.lower())
    return text.split()
````

````python
# --------------------------------------------
# 2. Remove stopwords (formal loop version)
# --------------------------------------------
def remove_stopwords(text):
    # Step 1: Tokenize the text
    tokens = tokenize(text)
    
    # Step 2: Create an empty list to store filtered words
    filtered_tokens = []
    
    # Step 3: Loop through each token
    for word in tokens:
        # Step 4: Check if the word is NOT in the stopwords list
        if word not in stop_words:
            # Step 5: Add the word to the filtered list
            filtered_tokens.append(word)
    
    # Step 6: Return the filtered list
    return filtered_tokens
````

````python
# --------------------------------------------
# 3. Practice with multiple sentences
# --------------------------------------------
for s in [sentence1, sentence2, sentence3]:
    print("Original:", s)
    print("Tokens:", tokenize(s))
    print("Without Stopwords:", remove_stopwords(s))
    print("-" * 40)
````

# NLP: Stemming vs Lemmatization

## Stemming
- **Definition:** A rule‑based process that chops off word endings to reduce words to their "stem."
- Example:  
  - "running" → "run"  
  - "studies" → "studi"  
- **Pros:** Fast, simple.  
- **Cons:** Can produce non‑dictionary forms ("studi" instead of "study").

---

## Lemmatization
- **Definition:** A dictionary‑based process that reduces words to their "lemma" (base form).
- Example:  
  - "running" → "run"  
  - "studies" → "study"  
- **Pros:** Produces valid words, more accurate.  
- **Cons:** Slower, requires linguistic resources.

---

## Key Insight
- **Stemming** is quick but crude.  
- **Lemmatization** is slower but linguistically correct.  
- Choice depends on task:  
  - Search engines → stemming (speed).  
  - Language understanding → lemmatization (accuracy).


# 📘 NLP – Stemming vs Lemmatization

## Stemming
- **Definition:** A rule‑based process that chops off word endings to reduce words to their "stem."
- **Examples (Porter Stemmer):**
  - running → run  
  - studies → studi  
  - flies → fli  
  - better → better  
  - cats → cat  
  - happiness → happi  
  - playing → play  
  - organization → organ  
  - quickly → quickli  
  - national → nation  

- **Pros:** Fast, simple.  
- **Cons:** Can produce non‑dictionary forms ("studi" instead of "study").

---

## Lemmatization
- **Definition:** A dictionary‑based process that reduces words to their "lemma" (base form).
- **Examples (WordNet Lemmatizer):**
  - running → run  
  - studies → study  
  - flies → fly  
  - better → good (adjective sense)  
  - cats → cat  
  - happiness → happiness  
  - playing → play  
  - organization → organization  
  - quickly → quickly  
  - national → national  

- **Pros:** Produces valid words, more accurate.  
- **Cons:** Slower, requires linguistic resources.

---

````python
# --------------------------------------------
# Stemming and Lemmatization in Python
# --------------------------------------------
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
````

````python
# Download WordNet (needed for lemmatization)
nltk.download("wordnet")
nltk.download("omw-1.4")
````

````python
# Example words
words = ["running", "studies", "flies", "better", "cats"]
````

````python
# --------------------------------------------
# 1. Stemming
# --------------------------------------------
stemmer = PorterStemmer()

for w in words:
    stemmed = stemmer.stem(w)
    print(f"Original: {w} -> Stemmed: {stemmed}")

print("-" * 40)
````

````python
# --------------------------------------------
# 2. Lemmatization
# --------------------------------------------
lemmatizer = WordNetLemmatizer()

for w in words:
    lemma = lemmatizer.lemmatize(w)  # Default POS = noun
    print(f"Original: {w} -> Lemma: {lemma}")

print("-" * 40)
````
