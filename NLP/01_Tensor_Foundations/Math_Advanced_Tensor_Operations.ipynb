#  Mathematical Operations in TensorFlow

### Element-wise Operations
Mathematical operations such as addition, subtraction, and multiplication are performed **element-wise**.  
This means that corresponding elements of the tensors are operated on individually.  
Element-wise operations are the foundation for scaling inputs, combining features, and applying activations.

---

### Broadcasting
When performing operations on tensors of different shapes, TensorFlow applies **broadcasting**.  
Broadcasting extends the smaller tensor along the missing dimensions to match the shape of the larger tensor.  
This allows flexible arithmetic without manual reshaping.

---

### Comparison Operations
TensorFlow supports **element-wise comparison** between tensors.  
The result is a tensor of boolean values (`True` or `False`) depending on whether the corresponding elements are equal.  
This is useful for filtering, masking, and conditional logic.

---

### Power Method
The **power operation** computes the value of one tensor raised to the power of another, performed element-wise.  
This is often used in normalization, polynomial transformations, and scaling.

---

### Matrix Multiplication
Matrix multiplication is central to deep learning.  
It involves multiplying rows of one matrix with columns of another and summing the products.  
Often, one of the matrices may be transposed to align dimensions.  
This operation underpins linear transformations in neural layers.

---

### Scalar Output
Some operations reduce tensors to a **scalar output**.  
For example, summation across all elements collapses dimensions into a single value.  
This reduction is critical for computing loss functions and aggregating results.

---

##  Key Insight
Mathematical operations in TensorFlow are not just computational tools, they are the **language of deep learning**.  

Every forward pass, loss calculation, and weight update is expressed through these operations, making them the theoretical backbone of how models learn and generalize.

```python
import tensorflow as tf
```

```python
a = tf.constant([1, 2, 3])
b = tf.constant([4, 5, 6])
```

```python
print("Addition:", tf.add(a, b))
```

```python
tf.subtract(a, b)
```

```python
tf.multiply(a, b)
```

### Broadcasting

```python
x = tf.constant([[1, 2, 3],
                 [4, 5, 6]])
y = tf.constant([1, 2, 3])  # smaller tensor
```

```python
tf.add(x, y)
```

## Comparison Operations

```python
a = tf.constant([1, 2, 3])
b = tf.constant([2, 2, 4])
```

```python
print("Equal:", tf.equal(a, b))   
print("Greater:", tf.greater(a, b)) 
```

```python
tf.less(a,b)
```

## Power Method

```python
base = tf.constant([2, 3, 4])
exp = tf.constant([3, 2, 1])
```

```python
print("Power:", tf.pow(base, exp))
```

## Matrix Multiplication

```python
A = tf.constant([[1, 2],
                 [3, 4]])
B = tf.constant([[5, 6],
                 [7, 8]])
```

```python
print("Matrix Multiplication:\n", tf.matmul(A, B))
```

## Scalar Output

```python
t = tf.constant([[1, 2, 3],
                 [4, 5, 6]])
```

```python
print("Sum of all elements:", tf.reduce_sum(t))   
```

```python
print("Mean of all elements:", tf.reduce_mean(t)) 
```
