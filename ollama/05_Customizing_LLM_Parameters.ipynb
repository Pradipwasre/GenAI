# Customizing LLM Output with Parameters

In this notebook, we explore how to **modify model behavior** using parameters in `ollama.generate()`.

---

### System Prompt
The **system prompt** defines the assistant’s role or personality.  
For example: *“You are a funny assistant”* makes the model respond humorously.

---

### Parameters with `options`
You can fine‑tune the model’s output style by passing parameters inside `options`.  
This allows you to control **creativity, diversity, and determinism** in the responses.

---

### Parameter Explanations

#### Temperature
- **Scale:** 0.0 → 1.0 (sometimes up to 2.0)
- **Meaning:** Controls randomness in token selection.
  - `0.0` → deterministic, always picks the most likely word.
  - `0.2–0.3` → low randomness, consistent outputs (good for JSON).
  - `0.7` → balanced, natural conversation.
  - `1.0+` → highly creative, but less predictable.

---

#### Top‑p (Nucleus Sampling)
- **Scale:** 0.0 → 1.0
- **Meaning:** Chooses from the smallest set of tokens whose cumulative probability ≥ *p*.
  - `0.5` → only consider the top 50% most likely tokens.
  - `1.0` → consider all tokens (no restriction).
- Lower values = more focused, higher values = more diverse.

---

#### Top‑k
- **Scale:** Integer (e.g., 1 → 100+)
- **Meaning:** Limits the number of candidate tokens to the top *k* most likely.
  - `top_k=1` → always pick the single most likely token.
  - `top_k=45` → consider the top 45 tokens before sampling.
- Smaller values = deterministic, larger values = more variety.

---

### Key Takeaway
- **Temperature** = creativity dial.  
- **Top‑p** = probability cutoff for diversity.  
- **Top‑k** = hard cap on candidate tokens.  

Together, these parameters let you **control precision vs creativity** in your model’s responses.

```python
import ollama

response = ollama.generate(
    model="llama3:latest",
    prompt="Tell me a joke about cricket",
    system="You are a funny assistant"
)

print(response.response)
```

```python
import ollama

response = ollama.generate(
    model="llama3:latest",
    prompt="Give me IPL hindi commentary in the style of Navjot Singh Sidhu",
    system="You are a funny assistant who speaks like Sidhu during cricket commentary"
)
```

```python
print(response.response)
```

```python
response = ollama.generate(
    model="llama3:latest",
    prompt="Give me IPL commentary in the style of Navjot Singh Sidhu",
    system="You are a funny assistant who speaks like Sidhu during cricket commentary",
    options={
        "temperature": 0.7,
        "top_p": 0.9,
        "top_k": 50
    }
)

print(response.response)
```
