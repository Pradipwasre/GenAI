## Ollama > Docs > API References > Generate a Completion

The `ollama.generate()` command is used to create text completions from a local model.  
Normally, it returns the entire response at once.  

### Parameter: `stream`
- `stream=True` allows you to receive the output **incrementally**, token by token (like ChatGPT typing word by word).
- Instead of a single dict, you get a **Python generator** that yields chunks of the response.
- This is useful for interactive apps, chatbots, or when you want to display text as it’s being generated.

With streaming, you can loop over the generator and print each chunk as it arrives.

```python
import ollama

# Generate a streaming response
stream_response = ollama.generate(
    model="llama3:latest",   # you can swap with deepseek-r1:latest or others
    prompt="Explain why summers feel hotter these days",
    stream=True              # enable streaming mode
)

# Print the generator object itself
print("Response object type:", type(stream_response))
```

```python
# Iterate over the stream and print chunks
print("\n Streaming Output:")
for chunk in stream_response:
    # Each chunk is a dict with 'response' key
    print(chunk['response'], end="", flush=True)
```

```python
print("\n\n Streaming complete")
```
