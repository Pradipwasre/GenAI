### Ollama GenerateResponse Output

- **Type**: `GenerateResponse` (dict-like object)
- **Contains**:
  - `model`: which model was used
  - `created_at`: timestamp
  - `done`: whether generation finished
  - `durations`: timing info for load, prompt, and eval
  - `prompt_eval_count` / `eval_count`: token counts
  - `response`: the actual generated text (main output)
  - `context`: token IDs (internal use)

The most important field for us is **`response`**, which contains the model’s answer:


```python
# Show just the generated text
print(response["response"])
```

```python
import ollama
import re

# Step 1: Generate response from Ollama
ollama_response = ollama.generate(
    model="llama3:latest",   # swap with deepseek-r1:latest or others
    prompt="Explain why summers feel hotter these days, include reasoning tags <think>...</think>"
)
```

```python
# Step 2: Extract the raw text output
raw_text = ollama_response.response
print("Raw Model Output:")
print(raw_text)
```

```python
# Step 3: Clean out <think> tags using regex
clean_text = re.sub(r"<think>.*?</think>", "", raw_text, flags=re.DOTALL)
print("\n Cleaned Output (without <think> tags):")
print(clean_text.strip())
```

```python
# Step 4: Show metadata for context
print("\nMetadata Info:")
print(f"Model Used: {ollama_response.model}")
print(f"Created At: {ollama_response.created_at}")
print(f"Tokens (Prompt): {ollama_response.prompt_eval_count}")
print(f"Tokens (Generated): {ollama_response.eval_count}")
print(f"Stop Reason: {ollama_response.done_reason}")
```

```python
# Step 5: Show type of response object
print("\n Response Object Type:")
print(type(ollama_response))  # <class 'ollama.responses.GenerateResponse'>
```
