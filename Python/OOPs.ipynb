# Introduction to Object-Oriented Programming (OOP) in Python

### 1. What is OOP?
- **Object-Oriented Programming (OOP)** is a programming paradigm that organizes code around **objects** rather than functions or logic alone.
- An **object** represents a real-world entity with **attributes (data)** and **methods (behavior)**.
- OOP helps make programs more **modular, reusable, and easier to maintain**.

---

### 2. Core Concepts of OOP
- **Classes and Objects**
  - A **class** is a blueprint or template that defines attributes and methods.
  - An **object** is an instance of a class, created from that blueprint.
  - Example in real life: A "Car" class defines properties like color and speed; each actual car is an object.

- **Encapsulation**
  - Bundling data (attributes) and behavior (methods) together inside a class.
  - Protects internal details and provides controlled access.

- **Inheritance**
  - Allows one class to **inherit properties and methods** from another.
  - Promotes code reuse and hierarchical relationships.

- **Polymorphism**
  - Means "many forms."
  - The same function or method name can behave differently depending on the object.
  - Example: A `speak()` method may produce different outputs for a "Dog" object vs. a "Cat" object.

- **Abstraction**
  - Hides complex implementation details and shows only the necessary features.
  - Helps simplify usage of classes and methods.

---

### 3. Why OOP Matters in Python
- Provides a **structured way** to design larger applications.
- Makes code **scalable** and **maintainable**.
- Encourages **reuse** of components across projects.
- Widely used in frameworks, libraries, and modern Python applications.

---

### 4. OOP in GenAI Context
- **Model Wrappers**: Classes are used to wrap GenAI models (e.g., defining methods for prompt creation, response handling).
- **Prompt Management**: Objects can represent prompts with attributes like text, context, and metadata.
- **Pipeline Design**: OOP helps structure GenAI workflows into modular components (e.g., preprocessing, inference, postprocessing).
- **Polymorphism in GenAI**:
  - Same method name (like `generate()`) can behave differently depending on the model (text generation, image generation, etc.).
- **Encapsulation for Security**:
  - Sensitive API keys and configurations are kept inside classes, ensuring controlled access.
- **Abstraction for Simplicity**:
  - Complex GenAI operations (tokenization, embeddings, API calls) are hidden behind simple class methods for learners and developers.

---

### 5. Key Takeaways
- OOP organizes code around **objects** with attributes and methods.
- Core principles: **Encapsulation, Inheritance, Polymorphism, Abstraction**.
- In GenAI, OOP is essential for building **modular, reusable, and secure workflows**.
- Understanding OOP prepares learners to design **custom methods, model wrappers, and pipelines** in GenAI projects.

## Defining a Class in Python

```python
# Defining a simple class in Python

class Student:
    # Class attribute
    institute = "CloudBlitz"
    
    # Constructor method to initialize attributes
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    # A method (behavior) inside the class
    def introduce(self):
        return f"My name is {self.name} and I am {self.age} years old."
```

```python
a = Student("Raj" , 19)
```

```python
a.institute
```

```python
class Student:
    # Class attribute
    institute = "CloudBlitz"
```

```python
# Creating an object (instance) of the Student class

a = Student()
```

```python
class Car:

    def __init__(self ,  brand , model , Rating):
        self.brand = brand
        self.model = model
        self.rating = Rating

    # Method 
    def full_name(self):
        return f"Brand Name: {self.brand} | Model Name: {self.model} | Ratings:  {self.rating}"
```

```python
a = Car("BMW" , "X6" , 5 )
```

```python
a.full_name()
```
