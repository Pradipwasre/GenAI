# Introduction to Object-Oriented Programming (OOP) in Python

### 1. What is OOP?
- **Object-Oriented Programming (OOP)** is a programming paradigm that organizes code around **objects** rather than functions or logic alone.
- An **object** represents a real-world entity with **attributes (data)** and **methods (behavior)**.
- OOP helps make programs more **modular, reusable, and easier to maintain**.

---

### 2. Core Concepts of OOP
- **Classes and Objects**
  - A **class** is a blueprint or template that defines attributes and methods.
  - An **object** is an instance of a class, created from that blueprint.
  - Example in real life: A "Car" class defines properties like color and speed; each actual car is an object.

- **Encapsulation**
  - Bundling data (attributes) and behavior (methods) together inside a class.
  - Protects internal details and provides controlled access.

- **Inheritance**
  - Allows one class to **inherit properties and methods** from another.
  - Promotes code reuse and hierarchical relationships.

- **Polymorphism**
  - Means "many forms."
  - The same function or method name can behave differently depending on the object.
  - Example: A `speak()` method may produce different outputs for a "Dog" object vs. a "Cat" object.

- **Abstraction**
  - Hides complex implementation details and shows only the necessary features.
  - Helps simplify usage of classes and methods.

---

### 3. Why OOP Matters in Python
- Provides a **structured way** to design larger applications.
- Makes code **scalable** and **maintainable**.
- Encourages **reuse** of components across projects.
- Widely used in frameworks, libraries, and modern Python applications.

---

### 4. OOP in GenAI Context
- **Model Wrappers**: Classes are used to wrap GenAI models (e.g., defining methods for prompt creation, response handling).
- **Prompt Management**: Objects can represent prompts with attributes like text, context, and metadata.
- **Pipeline Design**: OOP helps structure GenAI workflows into modular components (e.g., preprocessing, inference, postprocessing).
- **Polymorphism in GenAI**:
  - Same method name (like `generate()`) can behave differently depending on the model (text generation, image generation, etc.).
- **Encapsulation for Security**:
  - Sensitive API keys and configurations are kept inside classes, ensuring controlled access.
- **Abstraction for Simplicity**:
  - Complex GenAI operations (tokenization, embeddings, API calls) are hidden behind simple class methods for learners and developers.

---

### 5. Key Takeaways
- OOP organizes code around **objects** with attributes and methods.
- Core principles: **Encapsulation, Inheritance, Polymorphism, Abstraction**.
- In GenAI, OOP is essential for building **modular, reusable, and secure workflows**.
- Understanding OOP prepares learners to design **custom methods, model wrappers, and pipelines** in GenAI projects.

# Polymorphism in Object-Oriented Programming

## Definition
Polymorphism is a core concept in object-oriented programming that allows a single method name or interface to represent different behaviors depending on the object it is applied to. The term means "many forms."

## Key Points
- **Same method name, different behavior**: Different classes or objects can define methods with the same name, but each performs differently.
- **Flexibility**: Code can be written to work with different types of objects, as long as they share the same method signature.
- **Reusability**: Polymorphism reduces duplication by allowing one function to handle multiple object types.
- **Extensibility**: New classes can be added without changing existing code, as long as they implement the expected method.

## Types of Polymorphism
1. **Compile-time (Static)**: Achieved through method overloading or operator overloading.
2. **Run-time (Dynamic)**: Achieved through method overriding, where the method that gets executed depends on the object at runtime.

## Benefits
- Simplifies code by using a common interface.
- Makes systems easier to maintain and extend.
- Encourages design that is modular and adaptable.


````python
class SchoolStudent:
    def get_result(self):
        return "Result: Pass/Fail based on grades"

class CollegeStudent:
    def get_result(self):
        return "Result: GPA calculated"

class PhDStudent:
    def get_result(self):
        return "Result: Thesis evaluation"

def show_result(student):
    print(student.get_result())

# Polymorphism in action
show_result(SchoolStudent())   # Result: Pass/Fail based on grades
show_result(CollegeStudent())  # Result: GPA calculated
show_result(PhDStudent())      # Result: Thesis evaluation
````

````python
class PetrolCar:
    def fuel_type(self):
        return "Runs on Petrol"

class DieselCar:
    def fuel_type(self):
        return "Runs on Diesel"

class ElectricCar:
    def fuel_type(self):
        return "Runs on Electricity"
````

````python
# Polymorphism in action

def show_fuel(car):
    print(car.fuel_type())
````

````python
show_fuel(ElectricCar())
````

# Inheritance in Object-Oriented Programming

### Definition
Inheritance is a mechanism in object-oriented programming that allows one class (called the child or subclass) to acquire the properties and behaviors of another class (called the parent or superclass). It promotes code reuse and establishes a natural hierarchy between classes.

#### Key Points
- **Parent and Child Classes**: The child class inherits attributes and methods from the parent class.
- **Code Reusability**: Common functionality is defined once in the parent class and reused by child classes.
- **Extensibility**: Child classes can add new attributes or methods beyond what they inherit.
- **Overriding**: Child classes can redefine methods from the parent class to change or extend behavior.

#### Types of Inheritance
1. **Single Inheritance**: A child class inherits from one parent class.
2. **Multiple Inheritance**: A child class inherits from more than one parent class.
3. **Multilevel Inheritance**: A class inherits from a child class, forming a chain.
4. **Hierarchical Inheritance**: Multiple child classes inherit from the same parent class.
5. **Hybrid Inheritance**: A combination of different types of inheritance.

### Benefits
- Reduces code duplication.
- Improves maintainability by centralizing common logic.
- Supports polymorphism, allowing child classes to override or extend parent behavior.
- Encourages logical organization of

````python
# Parent class
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    
    def show_details(self):
        return f"Car: {self.brand} {self.model}"

# Child class inheriting from Car
class ElectricCar(Car):
    def __init__(self, brand, model, battery_capacity):
        # Call parent constructor
        super().__init__(brand, model)
        self.battery_capacity = battery_capacity
    
    # Overriding method
    def show_details(self):
        return f"Electric Car: {self.brand} {self.model}, Battery: {self.battery_capacity} kWh"

# Usage
car1 = Car("Toyota", "Corolla")
car2 = ElectricCar("Tesla", "Model 3", 75)

print(car1.show_details())   
print(car2.show_details())  
````

# Encapsulation in Object-Oriented Programming

### Definition
Encapsulation is the process of bundling data (attributes) and methods (functions) that operate on that data into a single unit called a class. It also restricts direct access to some components, which is a way of protecting the internal state of an object.

### Key Points
- **Data Hiding**: Internal details of a class are hidden from outside access. Only controlled methods (getters/setters) are exposed.
- **Access Modifiers**: In Python, naming conventions like `_protected` and `__private` are used to indicate restricted access.
- **Controlled Access**: Encapsulation ensures that data can only be modified in a controlled manner.
- **Security and Integrity**: Prevents accidental or unauthorized changes to the internal state of an object.

### Example (Conceptual)
Imagine a `Student` class:
- Attributes: `name`, `marks`
- Methods: `get_marks()`, `set_marks()`

Here, `marks` can be kept private, and only accessed or modified through the methods. This ensures that invalid data cannot be directly assigned.

## Benefits
- Protects the integrity of data.
- Simplifies maintenance by controlling how data is accessed and updated.
- Improves modularity by keeping implementation details hidden.
- Encourages cleaner and safer code design.

## Summary
Encapsulation is about **wrapping data and behavior together** and **restricting direct access** to safeguard the internal state of objects. It is one of the four pillars of OOP, alongside inheritance, polymorphism, and abstraction.

````python
class BankAccount:
    def __init__(self , owner , balance = 0):
        self.owner = owner
        self.__balance = balance

    def get_balance(self):
        return self.__balance
````

# Getter, Setter, and Normal Methods (Bank Example)

- **Setter Method** → when you **modify** a private variable (like `__balance`).  
  Example: `set_balance()` changes the value of `__balance`.

- **Getter Method** → when you **read or return** the private variable without changing it.  
  Example: `get_balance()` just returns the value of `__balance`.

- **Normal Method** → when you **use** the private variable for logic (like `if/else`, calculations, printing) but don’t directly change it.  

  Example: `withdraw()` checks `__balance` and decides if withdrawal is possible.

````python
class BankAccount:
    def __init__(self, owner , balance):
        self.owner = owner
        self.__balance = balance

    def get_balance(self):
        return self.__balance
    
    def deposit(self , amount):
        if amount > 0:
            self.__balance += amount
            return f"Deposited {amount} , New Balance : {self.__balance}"
    
    def withdraw(self, amount):
        if amount > 0:
            self.__balance -= amount
            return f"Withdrew {amount}, New Balance: {self.__balance}"
        
        else:
            return "Invalid Withdrawl amount"

    # Normal method (uses private variable but does not modify or directly return it)
    def check_status(self):
        if self.__balance >= 1000:
            return "Account status: Premium"
        else:
            return "Account status: Standard"

````

````python
user = BankAccount("Ajay" , 10000)
user.get_balance()
````

````python
user.withdraw(1000)
````

# How to Call and Use Getter, Setter, and Normal Methods (Bank Example)

### 1. Getter Method
- **Purpose**: To *read/display* the private variable safely.  
- **How to call**: Use the method name with `()` → `account.get_balance()`  
- **Result**: Returns the current balance without modifying it.

### 2. Setter Method
- **Purpose**: To *modify/update* the private variable in a controlled way.  
- **How to call**: Use the method name with `()` and pass a value → `account.deposit(500)` or `account.withdraw(200)`  
- **Result**: Changes the balance after validation.

### 3. Normal Method
- **Purpose**: To *use* the private variable in logic or calculations without directly setting or returning it.  
- **How to call**: Use the method name with `()` → `account.check_status()`  
- **Result**: Applies logic (like checking if balance is premium or standard) and returns a derived result.

---

### Example Usage
```python
account = BankAccount("Alice", 1000)

print(account.get_balance())    # Getter → shows balance
print(account.deposit(500))     # Setter → modifies balance
print(account.withdraw(200))    # Setter → modifies balance
print(account.check_status())   # Normal → uses balance in logic

# Abstract Methods in Python

## What is an Abstract Method?
An **abstract method** is a method that is declared but contains no implementation.  
It acts as a **rule or contract**: any class that inherits from the abstract base class must provide its own implementation of this method.

---

## Why Do We Use Abstract Methods?
- To **enforce consistency**: subclasses must implement certain methods.  
- To **define a blueprint**: senior developers can set rules that junior developers must follow.  
- To **separate design from implementation**: the abstract class defines *what* must be done, while subclasses decide *how* it is done.  
- To **support multiple variations**: different subclasses can implement the same abstract method in different ways.

---

## Real-World Analogy
Think of a **bank system**:
- The **main server (abstract class)** defines that every app must connect to the database and perform a security check.  
- The **abstract method (`security`)** is like a rule: *every app must have its own security process*.  
- Different apps implement this differently:
  - **Mobile App** → OTP + Fingerprint.  
  - **Web App** → Password + Captcha + 2FA.  
  - **ATM App** → Card + PIN.  

---

## Benefits of Abstract Methods
- **Uniformity**: ensures all subclasses follow the same essential rules.  
- **Flexibility**: allows different implementations depending on the platform or requirement.  
- **Scalability**: new login types (e.g., Face ID, Voice Authentication) can be added easily without changing the main blueprint.  
- **Maintainability**: keeps the system organized by separating shared features (concrete methods) from role-specific features (abstract methods).

---

## Summary
Abstract methods are used to **declare rules without implementation**.  
They guarantee that subclasses will implement those rules in their own way, making systems like bank applications **consistent, flexible, and easy to extend**.



````python
from abc import ABC, abstractmethod

# Senior developer creates abstract BankApp class
class BankApp(ABC):

    # Concrete method - has implementation
    def database(self):
        print("Connected to main bank server")

    # Abstract method - no implementation, just declaration
    @abstractmethod
    def security(self):
        pass


# Junior developer creates MobileApp class
class MobileApp(BankApp):

    def mobile_login(self):
        print("Login via Mobile App")

    # Real-world implementation of security
    def security(self):
        print("Security check: OTP + Fingerprint")


# Junior developer creates WebApp class
class WebApp(BankApp):

    def web_login(self):
        print("Login via Web Browser")

    # Real-world implementation of security
    def security(self):
        print("Security check: Password + Captcha + 2FA")


# Junior developer creates ATMApp class
class ATMApp(BankApp):

    def atm_login(self):
        print("Login via ATM Machine")

    # Real-world implementation of security
    def security(self):
        print("Security check: Card + PIN")
````

````python
# Usage
mobile = MobileApp()
web = WebApp()
atm = ATMApp()
````

````python
mobile.database()
mobile.mobile_login()
mobile.security()
````

````python
web.database()
web.web_login()
web.security()
````

````python
atm.database()
atm.atm_login()
atm.security()
````
