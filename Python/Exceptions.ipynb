# Error Handling in Python

## What is Error Handling?
Error handling in Python is the process of **detecting, managing, and responding to runtime errors** (exceptions) so that programs don’t crash unexpectedly.  
Instead of stopping execution when an error occurs, Python allows developers to **catch and handle exceptions gracefully**.

---

## Types of Errors
1. **Syntax Errors**
   - Occur when code violates Python’s grammar rules.
   - Example: `print "Hello"` (missing parentheses in Python 3).
   - Detected before the program runs.

2. **Runtime Errors (Exceptions)**
   - Occur while the program is running.
   - Example: dividing by zero, accessing a missing file, or using an undefined variable.

---

## Exception Hierarchy
- All exceptions in Python are derived from the **`BaseException`** class.
- Common built-in exceptions:
  - `ValueError` → invalid value.
  - `TypeError` → wrong data type.
  - `ZeroDivisionError` → division by zero.
  - `FileNotFoundError` → file not found.
  - `IndexError` → invalid list index.

---

## The `try-except` Block
- Used to catch and handle exceptions.
```python
try:
    risky_code()
except SomeException:
    handle_error()

## 1. ValueError

````python
try:
    num = int("abc")
except ValueError:
    print("Caught a ValueError")
````

## 2. ZeroDivisionError

````python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Caught a ZeroDivisionError")
````

## 3. TypeError

````python
try:
    result = "10" + 5
except TypeError:
    print("Caught a TypeError")
````

## 4. IndexError

````python
try:
    items = [1, 2, 3]
    print(items[5])
except IndexError:
    print("Caught an IndexError")
````

## 5. KeyError

````python
try:
    data = {"name": "Alice"}
    print(data["age"])
except KeyError:
    print("Caught a KeyError")
````

## 6. FileNotFoundError

````python
try:
    f = open("missing.txt")
except FileNotFoundError:
    print("Caught a FileNotFoundError")
````

## Exception 

````python
try:
    result = 10 / "two"   # risky code
except Exception as e:
    print("Caught an exception:", e)
````
