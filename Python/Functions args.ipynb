# Understanding `*args` in Python

## What is `*args`?
In Python, `*args` allows a function to accept a **variable number of positional arguments**.  
- The `*` operator collects all extra arguments into a **tuple**.  
- This makes functions flexible when the number of inputs is not fixed.  
- You can loop through `args`, apply conditions, or perform calculations.

---

## Why Use `*args`?
- **Positional arguments** are rigid: you must pass exactly the number of parameters defined.  
- **`*args`** solves this limitation by letting you pass **any number of arguments**.  
- Inside the function, `args` behaves like a tuple, so you can iterate, check conditions, or aggregate values.

---

## Example: Addition with `*args`
```python
def add_numbers(*args):
    return sum(args)

print(add_numbers(2, 4, 6))             # 12
print(add_numbers(1, 2, 3, 4, 5))       # 15
print(add_numbers(10, 20, 30, 40, 50))  # 150


## Example: Cricket Score Calculator with `*args`

```python
def cricket_scores(*runs):
    """
    Cricket score calculator using *args.
    Each argument represents the runs scored by a batsman.
    """
    total_runs = sum(runs)
    print("Cricket Match Summary")
    print(f"Players Scored: {len(runs)}")
    print(f"Total Team Runs: {total_runs}")

# Example usage
cricket_scores(45, 60, 30)

# Output:
# Cricket Match Summary
# Players Scored: 3
# Total Team Runs: 135

cricket_scores(100, 75, 50, 25, 10)
# Output:
# Cricket Match Summary
# Players Scored: 5
# Total Team Runs: 260


## Example: Optimized Blinkit Cart System with `*args`

```python
def blinkit_cart(*items):
    """
    Blinkit cart system using *args.
    Each item is passed as a tuple: (item_name, price).
    """
    print("Blinkit Cart Summary")
    total = 0
    
    # Loop through each tuple in items.
    # 'for name, price in items:' means:
    #   - Take each tuple (like ("Milk", 50))
    #   - Unpack it into variables: name = "Milk", price = 50
    for name, price in items:
        print(f"- {name}: ₹{price}")
        total += price
    
    print(f"Total Items: {len(items)}")
    print(f"Total Bill: ₹{total}")

# Example usage
blinkit_cart(("Milk", 50), ("Bread", 30), ("Eggs", 20))
# Output:
# Blinkit Cart Summary
# - Milk: ₹50
# - Bread: ₹30
# - Eggs: ₹20
# Total Items: 3
# Total Bill: ₹100

blinkit_cart(("Apples", 100), ("Bananas", 200), ("Oranges", 150), ("Juice", 75), ("Biscuits", 50))
# Output:
# Blinkit Cart Summary
# - Apples: ₹100
# - Bananas: ₹200
# - Oranges: ₹150
# - Juice: ₹75
# - Biscuits: ₹50
# Total Items: 5
# Total Bill: ₹575

````python
# Adding Student Marks

# Function with fixed 3 positional arguments
def add_three_marks(m1, m2, m3):
    return m1 + m2 + m3
````

````python
add_three_marks(85, 90, 78)
````

````python
# Function with *args (any number of marks)
def add_marks(*args):
    return sum(args)
````

````python
# Example usage
add_marks(85, 90, 78)
````

````python
# Function with fixed 3 subjects
def check_result(m1, m2, m3):
    if m1 >= 40 and m2 >= 40 and m3 >= 40:
        return "Pass"
    else:
        return "Fail"
````

````python
check_result(45, 55, 60)
````

````python
# Function with *args (any number of subjects)
def check_result(*marks):
    for mark in marks:
        if mark < 40:
            return "Fail"
    return "Pass"
````

````python
# Example usage
check_result(45, 55, 60)
````
