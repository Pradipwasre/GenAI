# 🔑 Understanding `**kwargs` in Python

## What is `**kwargs`?
- In Python, `**kwargs` stands for **keyword arguments**.
- It allows a function to accept a **variable number of keyword arguments**.
- The `**` operator collects these arguments into a **dictionary**:
  - Keys → the argument names
  - Values → the argument values
- This makes functions **flexible and extensible**, especially when you don’t know in advance which named parameters will be passed.

---

## Key Characteristics
- **Positional arguments**: Fixed order, must be passed exactly as defined.
- **`*args`**: Collects extra positional arguments into a tuple.
- **`**kwargs`**: Collects extra keyword arguments into a dictionary.
- Inside the function, you can access `kwargs` just like a normal dictionary (`kwargs.items()`, `kwargs.keys()`, `kwargs.values()`).

---

## Why Use `**kwargs`?
- To handle **optional or dynamic parameters** without changing the function signature.
- To make functions more **general-purpose** and **future-proof**.
- To allow passing **configuration options, metadata, or flexible attributes**.
- Commonly used in:
  - API design (handling flexible parameters)
  - Class constructors (passing dynamic attributes)
  - Wrappers and decorators (forwarding arguments)

---

## Important Notes
- `**kwargs` must always come **after positional and `*args`** in the function definition.
- The order of parameters in a function definition is:
  1. Positional arguments
  2. `*args`
  3. `**kwargs`
- This ensures Python knows how to separate fixed, variable positional, and variable keyword arguments.

---

## Summary
- `**kwargs` = "keyword arguments" → collects into a dictionary.
- Use it when you want your function to accept **flexible named inputs**.
- It complements positional arguments and `*args`, giving you full control over how functions receive data.
- With `**kwargs`, your functions become **adaptable, reusable, and scalable**.

# 🎓 Understanding `**kwargs` in Python with a College Example

## What is `**kwargs`?
- In Python, `**kwargs` stands for **keyword arguments**.
- It allows a function to accept a **variable number of named arguments**.
- These arguments are collected into a **dictionary**:
  - Keys → the argument names
  - Values → the argument values
- This makes functions **flexible** when you don’t know in advance which attributes or details will be passed.

```python
def student_profile(name, **kwargs):
    """
    Student profile system using positional + **kwargs.
    'name' is a fixed positional argument.
    '**kwargs' captures flexible details about the student.
    """
    print(f"Student Name: {name}")
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
student_profile("Sachin" , Age = 30 , Course = "Data Science")
```

```python
def student_profile(name, **kwargs):
    """
    Student profile system using positional + **kwargs.
    'name' is a fixed positional argument.
    '**kwargs' captures flexible details about the student.
    Instead of printing, we store the details in a dictionary.
    """
    profile = {"name": name}
    profile.update(kwargs)
    return profile
```

```python
student1 = student_profile("Pradip", age=20, course="Computer Science", year="2nd Year", hostel="A-Block")
student2 = student_profile("Sneha", age=21, course="Data Science", year="Final Year")
```

```python
## Example: Car Details with `**kwargs`


def car_details(brand, **kwargs):
    """
    Car details system using positional + **kwargs.
    'brand' is a fixed positional argument.
    '**kwargs' captures flexible attributes of the car.
    """
    print(f"Car Brand: {brand}")
    for key, value in kwargs.items():
        print(f"{key}: {value}")

# Example usage
car_details("Mercedes", model="C-Class", year=2022, color="Black", price="₹50,00,000")
```

```python
car_details("Tesla", model="Model S", year=2023, color="Red", autopilot=True)
```
